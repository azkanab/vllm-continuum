# DAGShare analysis

Thin wrapper over `dagshare_exp/analyze.py`. Set `RESULTS_ROOT` to a fetched results directory (default: synthetic fixtures). Definitions are in the module docstring.

```python
import sys
import tempfile
from pathlib import Path

REPO = Path.cwd().resolve().parent if Path.cwd().name == "dagshare_exp" else Path.cwd().resolve()
sys.path.insert(0, str(REPO))

import pandas as pd
from IPython.display import Image, display
from dagshare_exp import analyze

RESULTS_ROOT = REPO / "dagshare_exp" / "fixtures" / "results"
OUT_DIR = Path(tempfile.gettempdir()) / "dagshare_analysis"
pd.set_option("display.max_columns", 80)
```

```python
res = analyze.run(RESULTS_ROOT, OUT_DIR)
print(analyze.__doc__)
```
